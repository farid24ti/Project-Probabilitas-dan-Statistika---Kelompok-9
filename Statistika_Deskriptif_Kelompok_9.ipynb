**Tabel 17 Statistika**

```python
hasil = pd.DataFrame(index=[
    'Mean',
    'Median',
    'Modus',
    'Minimum',
    'Maximum',
    'Range',
    'Varians',
    'Standar Deviasi',
    'Q1',
    'Q2',
    'Q3',
    'P10',
    'P90',
    'D1',
    'D5',
    'D9',
    'IQR',
    'Skewness',
    'Kurtosis',
    'CV'
], columns=numeric_cols)
```

```python
hasil.loc['Mean'] = df[numeric_cols].mean()
hasil.loc['Median'] = df[numeric_cols].median()
hasil.loc['Modus'] = df[numeric_cols].mode().iloc[0]
hasil.loc['Minimum'] = df[numeric_cols].min()
hasil.loc['Maximum'] = df[numeric_cols].max()
hasil.loc['Range'] = df[numeric_cols].max() - df[numeric_cols].min()
hasil.loc['Varians'] = df[numeric_cols].var()
hasil.loc['Standar Deviasi'] = df[numeric_cols].std()

hasil.loc['Q1'] = df[numeric_cols].quantile(0.25)
hasil.loc['Q2'] = df[numeric_cols].quantile(0.50)
hasil.loc['Q3'] = df[numeric_cols].quantile(0.75)

hasil.loc['P10'] = df[numeric_cols].quantile(0.10)
hasil.loc['P90'] = df[numeric_cols].quantile(0.90)

hasil.loc['D1'] = df[numeric_cols].quantile(0.10)
hasil.loc['D5'] = df[numeric_cols].quantile(0.50)
hasil.loc['D9'] = df[numeric_cols].quantile(0.90)

hasil.loc['IQR'] = (
    df[numeric_cols].quantile(0.75)
    - df[numeric_cols].quantile(0.25)
)

hasil.loc['Skewness'] = df[numeric_cols].skew()
hasil.loc['Kurtosis'] = df[numeric_cols].kurt()

hasil.loc['CV'] = (
    df[numeric_cols].std()
    / df[numeric_cols].mean()
) * 100
```

```python
hasil
```

**Frekuensi Python**

```python
# brand
frekuensi_brand = df['brand'].value_counts()

frekuensi_brand
```

```python
# persentase
persentase_brand = (
    df['brand']
    .value_counts(normalize=True)
    * 100
)

persentase_brand
```

```python
brand_table = pd.DataFrame({
    'Frekuensi': frekuensi_brand,
    'Persentase (%)': persentase_brand
})

brand_table
```

```python
# transmission
transmission_table = pd.DataFrame({
    'Frekuensi': df['transmission'].value_counts(),
    'Persentase (%)':
        df['transmission'].value_counts(normalize=True) * 100
})

transmission_table
```

**Visualisasi Python**

```python
# histogram harga
plt.figure(figsize=(10,6))

plt.hist(df['price (Rp)'], bins=20)

plt.title('Distribusi Harga Mobil Bekas')
plt.xlabel('Harga (Rp)')
plt.ylabel('Frekuensi')

plt.show()
```

```python
# histogram mileage
plt.figure(figsize=(10,6))

plt.hist(df['mileage (km)'], bins=20)

plt.title('Distribusi Jarak Tempuh Mobil')
plt.xlabel('Mileage (km)')
plt.ylabel('Frekuensi')

plt.show()
```

```python
# mean vs median harga
mean_price = df['price (Rp)'].mean()
median_price = df['price (Rp)'].median()

plt.figure(figsize=(8,5))

plt.bar(
    ['Mean', 'Median'],
    [mean_price, median_price]
)

plt.title('Perbandingan Mean dan Median Harga Mobil')
plt.ylabel('Harga (Rp)')

plt.show()
```

**Frekuensi Brand**

```python
brand_freq = df['brand'].value_counts()

plt.figure(figsize=(10,6))

brand_freq.sort_values().plot(kind='barh')

plt.title('Frekuensi Merek Mobil')
plt.xlabel('Jumlah Kendaraan')
plt.ylabel('Brand')

plt.show()
```
