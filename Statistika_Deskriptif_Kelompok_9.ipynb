```python
data = pd.read_csv("used_car_kel9.csv", sep=";")

print(data)
```

```python
print("Jumlah baris :", data.shape[0])
print("Jumlah kolom:", data.shape[1])
```

```python
missing = data.isnull().sum()

print(missing)
```

```python
print("Total missing value:", data.isnull().sum().sum())
```

```python
data["mileage_km"] = data["mileage (km)"] * 1000
```

```python
variabel_numerik = [
    "mileage_km",
    "price (Rp)",
    "instalment (Rp|Monthly)"
]

data_numerik = data[variabel_numerik]

print(data_numerik.head())
```

```python
statistik = data_numerik.describe()

print(statistik)
```

```python
statistik = data_numerik.describe().T

statistik["median"] = data_numerik.median()
statistik["mode"] = data_numerik.mode().iloc[0]
statistik["variance"] = data_numerik.var()
statistik["range"] = data_numerik.max() - data_numerik.min()
statistik["standard_error"] = data_numerik.sem()
statistik["skewness"] = data_numerik.skew()
statistik["kurtosis"] = data_numerik.kurtosis()

print(statistik)
```

```python
plt.figure(figsize=(8, 5))

plt.boxplot(
    data["price (Rp)"],
    vert=False,
    patch_artist=True
)

plt.title("Boxplot Harga Mobil Bekas")
plt.xlabel("Harga (Rp)")

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.boxplot(
    data["mileage_km"],
    vert=False,
    patch_artist=True
)

plt.title("Boxplot Jarak Tempuh Mobil")
plt.xlabel("Mileage (km)")

plt.tight_layout()
plt.show()
```

```python
automatic = data[data["transmission"] == "Automatic"]["price (Rp)"]
manual = data[data["transmission"] == "Manual"]["price (Rp)"]

plt.figure(figsize=(8, 5))

plt.violinplot(
    [automatic, manual],
    showmeans=True,
    showmedians=True
)

plt.xticks(
    [1, 2],
    ["Automatic", "Manual"]
)

plt.title("Distribusi Harga Berdasarkan Jenis Transmisi")
plt.xlabel("Transmission")
plt.ylabel("Harga (Rp)")

plt.tight_layout()
plt.show()
```

```python
fitur = [
    "rear camera",
    "sun roof",
    "auto retract mirror",
    "electric parking brake",
    "map navigator",
    "vehicle stability control",
    "keyless push start",
    "sports mode",
    "360 camera view",
    "power sliding door",
    "auto cruise control"
]

jumlah_fitur = data[fitur].sum().sort_values(ascending=True)

plt.figure(figsize=(9, 6))

plt.barh(
    jumlah_fitur.index,
    jumlah_fitur.values
)

plt.title("Jumlah Mobil Berdasarkan Fitur")
plt.xlabel("Jumlah Mobil")
plt.ylabel("Fitur")

plt.tight_layout()
plt.show()
```
