```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from google.colab import files

uploaded = files.upload()          # pilih used_car_kel9.csv
data = pd.read_csv("used_car_kel9.csv", sep=",")
data["mileage_km"] = data["mileage (km)"] * 1000
print(data.shape)
```

```python
from scipy import stats
import statsmodels.api as sm
from statsmodels.stats.outliers_influence import variance_inflation_factor

pd.set_option("display.width", 200)
pd.set_option("display.max_columns", None)

print("Pustaka tambahan berhasil diimpor")
```

```python
variabel_korelasi = [
    "year",
    "mileage_km",
    "price (Rp)",
    "instalment (Rp|Monthly)"
]

data_korelasi = data[variabel_korelasi]

print(data_korelasi.head())
print()
print("Jumlah baris :", data_korelasi.shape[0])
print("Jumlah kolom :", data_korelasi.shape[1])
```

```python
korelasi_pearson = data_korelasi.corr(method="pearson")

print(korelasi_pearson.round(4))
```

```python
korelasi_spearman = data_korelasi.corr(method="spearman")

print(korelasi_spearman.round(4))
```

```python
harga = data["price (Rp)"]
hasil = []

for kolom in ["mileage_km", "year", "instalment (Rp|Monthly)"]:
    r, p_r = stats.pearsonr(data[kolom], harga)
    rho, p_rho = stats.spearmanr(data[kolom], harga)
    hasil.append([kolom, r, p_r, rho, p_rho])

tabel_korelasi = pd.DataFrame(
    hasil,
    columns=["variabel", "pearson_r", "p_pearson", "spearman_rho", "p_spearman"]
)
tabel_korelasi["signifikan_pearson"] = np.where(tabel_korelasi["p_pearson"] < 0.05, "Ya", "Tidak")
tabel_korelasi["signifikan_spearman"] = np.where(tabel_korelasi["p_spearman"] < 0.05, "Ya", "Tidak")

print(tabel_korelasi.to_string(index=False, float_format=lambda v: f"{v:.4g}"))
```

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

pasangan = [("mileage_km", "Jarak Tempuh (km)"), ("year", "Tahun Pembuatan")]

for ax, (kolom, label) in zip(axes, pasangan):
    x = data[kolom]
    y_harga = data["price (Rp)"]
    ax.scatter(x, y_harga, alpha=0.4, s=15)

    kemiringan, intersep = np.polyfit(x, y_harga, 1)
    garis_x = np.linspace(x.min(), x.max(), 100)
    ax.plot(garis_x, kemiringan * garis_x + intersep, color="red")

    ax.set_title(f"{label} vs Harga (r = {x.corr(y_harga):.4f})")
    ax.set_xlabel(label)
    ax.set_ylabel("Harga (Rp)")

plt.tight_layout()
plt.show()
```

```python
nama_singkat = ["Tahun", "Jarak tempuh", "Harga", "Cicilan"]

fig, axes = plt.subplots(1, 2, figsize=(14, 5.5), constrained_layout=True)

for ax, matriks, judul in zip(axes, [korelasi_pearson, korelasi_spearman], ["Pearson", "Spearman"]):
    gambar = ax.imshow(matriks.values, cmap="coolwarm", vmin=-1, vmax=1)
    ax.set_xticks(range(4))
    ax.set_xticklabels(nama_singkat, rotation=45, ha="right")
    ax.set_yticks(range(4))
    ax.set_yticklabels(nama_singkat)
    for i in range(4):
        for j in range(4):
            nilai = matriks.values[i, j]
            ax.text(j, i, f"{nilai:.2f}", ha="center", va="center",
                    color="white" if abs(nilai) > 0.6 else "black")
    ax.set_title(f"Matriks Korelasi {judul}")

fig.colorbar(gambar, ax=axes, shrink=0.8)
plt.show()
```

```python
X = data[["mileage_km", "year"]]
y = data["price (Rp)"]

X = sm.add_constant(X)

print(X.head())
print()
print("Jumlah data :", len(X))
```

```python
model = sm.OLS(y, X).fit()

print(model.summary())
```

```python
tabel_koefisien = pd.DataFrame({
    "koefisien": model.params,
    "galat_baku": model.bse,
    "t_hitung": model.tvalues,
    "p_value": model.pvalues
})
tabel_koefisien[["ci95_bawah", "ci95_atas"]] = model.conf_int()

print(tabel_koefisien.to_string(float_format=lambda v: f"{v:.6g}"))
print()

rmse = np.sqrt(np.mean(model.resid ** 2))
mae = np.mean(np.abs(model.resid))

print("R-squared          :", round(model.rsquared, 4))
print("Adjusted R-squared :", round(model.rsquared_adj, 4))
print("F-statistic        :", round(model.fvalue, 2), "| p-value F =", f"{model.f_pvalue:.3e}")
print(f"RMSE               : Rp{rmse:,.0f}")
print(f"MAE                : Rp{mae:,.0f}")
print()

b0, b1, b2 = model.params
print(f"Harga = {b0:,.0f} + {b1:,.2f} x mileage_km + {b2:,.0f} x year")
```

```python
vif = pd.DataFrame({
    "variabel": X.columns,
    "VIF": [variance_inflation_factor(X.values, i) for i in range(X.shape[1])]
})

print(vif[vif["variabel"] != "const"].round(4).to_string(index=False))
```

```python
fig, axes = plt.subplots(1, 3, figsize=(16, 4.5))

# Harga aktual vs harga prediksi
axes[0].scatter(model.fittedvalues, y, alpha=0.4, s=15)
batas = [y.min(), y.max()]
axes[0].plot(batas, batas, color="red")
axes[0].set_title("Harga Aktual vs Prediksi")
axes[0].set_xlabel("Harga prediksi (Rp)")
axes[0].set_ylabel("Harga aktual (Rp)")

# Residual vs harga prediksi
axes[1].scatter(model.fittedvalues, model.resid, alpha=0.4, s=15)
axes[1].axhline(0, color="red")
axes[1].set_title("Residual vs Harga Prediksi")
axes[1].set_xlabel("Harga prediksi (Rp)")
axes[1].set_ylabel("Residual (Rp)")

# Histogram residual
axes[2].hist(model.resid, bins=30, edgecolor="black")
axes[2].set_title("Histogram Residual")
axes[2].set_xlabel("Residual (Rp)")
axes[2].set_ylabel("Jumlah mobil")

plt.tight_layout()
plt.show()

w, p_shapiro = stats.shapiro(model.resid)
print(f"Uji Shapiro-Wilk residual: W = {w:.4f}, p-value = {p_shapiro:.3e}")
```

```python
data_baru = pd.DataFrame({
    "mileage_km": [10000, 30000, 60000, 100000],
    "year": [2022, 2020, 2018, 2015]
})

data_baru = sm.add_constant(data_baru, has_constant="add")
data_baru["prediksi_harga (Rp)"] = model.predict(data_baru[["const", "mileage_km", "year"]])

print(data_baru.to_string(index=False, float_format=lambda v: f"{v:,.0f}"))
```
